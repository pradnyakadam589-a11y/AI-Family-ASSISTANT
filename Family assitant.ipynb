```python

# AI Family Assistant System
# BCA 2nd Year AI Project

family_tasks = []
family_events = []

def show_menu():
    print("\n===== AI FAMILY ASSISTANT =====")
    print("1. Ask Family Assistant")
    print("2. Add Family Task")
    print("3. View Family Tasks")
    print("4. Add Family Event")
    print("5. View Family Events")
    print("6. Get Family Activity Suggestion")
    print("7. Exit")

def chatbot():
    question = input("\nAsk your assistant: ").lower()

    if "hello" in question or "hi" in question:
        print("AI: Hello! Welcome to your Family Assistant.")

    elif "task" in question:
        print("AI: You can add and view tasks from the main menu.")

    elif "birthday" in question:
        print("AI: Check your family events for birthday reminders.")

    elif "study" in question:
        print("AI: Make a study timetable and take short breaks.")

    elif "activity" in question or "sunday" in question:
        print("AI: You could enjoy a family game, picnic, or movie.")

    elif "help" in question:
        print("AI: I can help with tasks, events, study, and activities.")

    else:
        print("AI: Sorry, I don't know that yet. Try asking about tasks, study, or activities.")

def add_task():
    name = input("Enter family member name: ")
    task = input("Enter task: ")
    family_tasks.append({"member": name, "task": task})
    print("Task added successfully!")

def view_tasks():
    print("\n--- FAMILY TASKS ---")

    if not family_tasks:
        print("No tasks available.")
    else:
        for i, item in enumerate(family_tasks, start=1):
            print(f"{i}. {item['member']}: {item['task']}")

def add_event():
    event = input("Enter event name: ")
    date = input("Enter event date (DD-MM-YYYY): ")

    family_events.append({"event": event, "date": date})
    print("Family event added successfully!")

def view_events():
    print("\n--- FAMILY EVENTS ---")

    if not family_events:
        print("No events available.")
    else:
        for i, item in enumerate(family_events, start=1):
            print(f"{i}. {item['event']} - {item['date']}")

def suggest_activity():
    print("\n--- FAMILY ACTIVITY SUGGESTIONS ---")
    mood = input("Choose mood (relaxing, active, learning): ").lower()

    if mood == "relaxing":
        print("AI Suggestion: Watch a movie or enjoy a family game.")
    elif mood == "active":
        print("AI Suggestion: Go for a walk or play an outdoor game.")
    elif mood == "learning":
        print("AI Suggestion: Read together or try a family quiz.")
    else:
        print("AI Suggestion: Spend quality time together!")

while True:
    show_menu()
    choice = input("Enter your choice (1-7): ")

    if choice == "1":
        chatbot()
    elif choice == "2":
        add_task()
    elif choice == "3":
        view_tasks()
    elif choice == "4":
        add_event()
    elif choice == "5":
        view_events()
    elif choice == "6":
        suggest_activity()
    elif choice == "7":
        print("Thank you for using AI Family Assistant!")
        break
    else:
        print("Invalid choice. Please try again.")
```
